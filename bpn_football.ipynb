```python
import os
import pandas as pd

os.makedirs("data_football", exist_ok=True)

csv_path = "data_football/players_21.csv"

if not os.path.exists(csv_path):
    os.system("kaggle datasets download stefanoleone992/fifa-21-complete-player-dataset -f players_21.csv -p data_football --unzip")
    print("Download complete.")
else:
    print("File already exists, skipping download.")
```

```python
identifier = "short_name"
target = "preferred_foot"
features = [
    "age", "height_cm", "weight_kg",
    "pace", "shooting", "passing", "dribbling", "defending", "physic",
    "attacking_crossing", "attacking_finishing", "attacking_heading_accuracy",
    "attacking_short_passing", "attacking_volleys",
    "skill_dribbling", "skill_curve", "skill_long_passing", "skill_ball_control",
    "movement_acceleration", "movement_sprint_speed", "movement_agility",
    "movement_reactions", "movement_balance",
    "power_shot_power", "power_jumping", "power_stamina", "power_strength",
    "mentality_aggression", "mentality_positioning", "mentality_vision",
    "defending_marking", "defending_standing_tackle",
    "goalkeeping_diving", "goalkeeping_reflexes"
]

# Load only the necessary columns
df = pd.read_csv("data_football/players_21.csv", usecols=[identifier, target] + features, low_memory=False)

# Drop rows with missing values
df = df.dropna()

print(f"Shape: {df.shape}")
df.head()
```
